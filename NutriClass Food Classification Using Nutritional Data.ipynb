# To find outlier 

```python
def cap_outliers(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Count how many outliers exist before capping
    outlier_count = ((data[col] < lower) | (data[col] > upper)).sum()
    print(f"{col}: {outlier_count} outliers found (capped to [{lower:.2f}, {upper:.2f}])")

    # Cap instead of removing — keeps the row, just limits the extreme value
    data[col] = np.where(data[col] < lower, lower, data[col]) # numpy.where(condition, x, y) that means np.array(condition, True, False)is used for conditional selection and replacement in NumPy arrays.
    data[col] = np.where(data[col] > upper, upper, data[col]) # numpy.where(condition, x, y) that means np.array(condition, True, False)is used for conditional selection and replacement in NumPy arrays.
    return data

for col in ['Calories', 'Protein', 'Carbs', 'Fat']:
    Food = cap_outliers(Food_2, col)

```

# Hyperparameter Tuning

```python
from sklearn.pipeline import make_pipeline
```

```python
train_scores = [] # For training
val_scores = [] # For Tuning 


scaler = StandardScaler()

for la in np.arange(0.01, 5000.0, 100): # range of values of Lambda
  scaled_lr = make_pipeline(scaler, LogisticRegression(C=1/la))
  scaled_lr.fit(X_train, y_train)
  train_score = scaled_lr.score(X_train,y_train)
  val_score = scaled_lr.score(X_val,y_val)

  train_scores.append(train_score)
  val_scores.append(val_score)


```

```python
plt.figure(figsize=(10,5))
plt.plot(list(np.arange(0.01, 5000.0, 100)), train_scores, label="train")
plt.plot(list(np.arange(0.01, 5000.0, 100)), val_scores, label="val")
plt.legend(loc='lower right')

plt.xlabel("Regularization Parameter(λ)")
plt.ylabel("Accuracy")
plt.grid()
plt.show()
```

```python
def training(model,X_train,y_train,X_test,y_test):
  model.fit(X_train,y_train)

  y_pred = model.predict(X_train)
  f1_train = f1_score(y_train,y_pred)

  y_pred = model.predict(X_test)
  f1_test = f1_score(y_test,y_pred)
  return f1_train,f1_test
```
